```python
LAST_NAME = "CAJIDA"
SEED_NUM = 3
FAVORITE_ARTIST = "TUKI."
```

```python
student_names = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

scores = []
for i, char in enumerate(FAVORITE_ARTIST):
    score = (ord(char) + SEED_NUM * (i + 1)) % 101
    scores.append(score)

while len(scores) < 25:
    scores.append((scores[len(scores) % len(scores)] + SEED_NUM) % 101)
scores = scores[:25]

print("Student Names:", student_names)
print("Subjects:", subjects)
print("Generated Scores:", scores)
```

```python
records = []
idx = 0
for student in student_names:
    for subject in subjects:
        records.append({"student": student, "subject": subject, "score": scores[idx]})
        idx += 1

records.append(dict(records[0]))
records.append({"student": student_names[2], "subject": subjects[3], "score": None})

print(f"Total Records Generated: {len(records)}")
```

```python
def classify_score(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 75:
        return "GOOD"
    elif score >= 60:
        return "AVERAGE"
    else:
        return "NEEDS IMPROVEMENT"

for r in records:
    r["category"] = classify_score(r["score"])

print("Sample Classified Record:", records[0])
```

```python
seen = set()
repeated = []
for r in records:
    key = (r["student"], r["subject"])
    if key in seen:
        repeated.append(r)
    seen.add(key)

incomplete = [r for r in records if r["score"] is None]

print("Repeated Records:", repeated)
print("Incomplete Records:", incomplete)
```

```python
from collections import defaultdict

student_totals = defaultdict(list)
for r in records:
    if r["score"] is not None:
        student_totals[r["student"]].append(r["score"])
student_summary = {s: round(sum(v)/len(v), 2) for s, v in student_totals.items()}

subject_totals = defaultdict(list)
for r in records:
    if r["score"] is not None:
        subject_totals[r["subject"]].append(r["score"])
subject_summary = {s: round(sum(v)/len(v), 2) for s, v in subject_totals.items()}

ordered_results = sorted(records, key=lambda r: (r["student"], r["subject"]))

print("\n=== CONSOLIDATED PERFORMANCE REPORT ===")
print(f"Total Records: {len(records)}")
print(f"Student Averages: {student_summary}")
print(f"Subject Averages: {subject_summary}")
```
