```python
LAST_NAME = "CAJIDA"
SEED_NUM = 3
FAVORITE_ARTIST = "TUKI."
```

```python
customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {"NOTEBOOK": 50, "PEN": 15, "USB": 250, "HEADPHONES": 800, "CALCULATOR": 120}

quantities = []
for i, char in enumerate(FAVORITE_ARTIST):
    qty = (ord(char) + SEED_NUM) % 10 + 1
    quantities.append(qty)

while len(quantities) < 6:
    quantities.append((quantities[len(quantities) % len(quantities)] + SEED_NUM) % 10 + 1)
quantities = quantities[:6]

print("Customers:", customers)
print("Products:", products)
print("Quantities:", quantities)
```

```python
transactions = []
for i in range(5):
    transactions.append({
        "customer": customers[i],
        "product": products[i],
        "quantity": quantities[i]
    })

transactions.append(dict(transactions[0]))

print("Generated Transaction Records:", transactions)
```

```python
for t in transactions:
    t["value"] = t["quantity"] * prices[t["product"]]

total_value = sum(t["value"] for t in transactions)

seen = set()
repeated = []
distinct = []
for t in transactions:
    key = (t["customer"], t["product"])
    if key in seen:
        repeated.append(t)
    else:
        distinct.append(t)
    seen.add(key)

print(f"Total Transaction Value: {total_value}")
print(f"Repeated Transactions: {repeated}")
print(f"Distinct Transactions: {distinct}")
```

```python
from collections import defaultdict

customer_totals = defaultdict(int)
product_totals = defaultdict(int)
for t in transactions:
    customer_totals[t["customer"]] += t["value"]
    product_totals[t["product"]] += t["value"]

ordered_results = sorted(transactions, key=lambda t: (t["customer"], t["product"]))

print(f"Customer Summaries: {dict(customer_totals)}")
print(f"Product Summaries: {dict(product_totals)}")
print(f"\n=== FINAL CONSOLIDATED TRANSACTION REPORT ===")
print(f"Total Transactions: {len(transactions)} | Total Value: {total_value}")
```
