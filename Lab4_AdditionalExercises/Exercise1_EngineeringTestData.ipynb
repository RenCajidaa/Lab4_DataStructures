```python
LAST_NAME = "CAJIDA"
SEED_NUM = 3
FAVORITE_ARTIST = "TUKI."

test_sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

values = []
for i, char in enumerate(FAVORITE_ARTIST):
    val = (ord(char) + SEED_NUM * (i + 1)) % 101
    values.append(val)
while len(values) < 25:
    values.append((values[len(values) % len(values)] + SEED_NUM) % 101)
values = values[:25]

print("Test Sources:", test_sources)
print("Categories:", categories)
print("Generated Values:", values)
```

```python
records = []
idx = 0
for source in test_sources:
    for cat in categories:
        records.append({"source": source, "category": cat, "value": values[idx]})
        idx += 1

# Repeated test-source/category record
records.append(dict(records[0]))

# Additional measurements under at least two categories
records.append({"source": test_sources[2], "category": categories[0], "value": (values[2] + SEED_NUM) % 101})
records.append({"source": test_sources[3], "category": categories[1], "value": (values[3] + SEED_NUM) % 101})

print(f"Total Records Generated: {len(records)}")
```

```python
from collections import defaultdict

seen = {}
repeated = []
distinct_keys = set()
for r in records:
    key = (r["source"], r["category"])
    if key in seen:
        repeated.append(r)
    else:
        distinct_keys.add(key)
    seen[key] = seen.get(key, 0) + 1

print("Repeated Records:", repeated)
print("Distinct Source/Category Combinations:", len(distinct_keys))
```

```python
category_totals = defaultdict(list)
source_totals = defaultdict(list)
for r in records:
    category_totals[r["category"]].append(r["value"])
    source_totals[r["source"]].append(r["value"])

category_summary = {c: round(sum(v)/len(v), 2) for c, v in category_totals.items()}
source_summary = {s: round(sum(v)/len(v), 2) for s, v in source_totals.items()}

ordered_results = sorted(records, key=lambda r: (r["source"], r["category"]))

print("Category Averages:", category_summary)
print("Source Averages:", source_summary)
```

```python
print("\n=== FINAL ORGANIZED AND SUMMARIZED TEST DATASET ===")
print(f"Total Records: {len(records)}")
print(f"Category Averages: {category_summary}")
print(f"Source Averages: {source_summary}")
```
